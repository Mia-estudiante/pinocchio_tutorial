```python
import os
from pathlib import Path
from sys import argv

import pinocchio as pin
from pinocchio.robot_wrapper import RobotWrapper
```

```python
pinocchio_model_dir = Path(__file__).parent
model_path = Path((pinocchio_model_dir / "example-robot-data/robots")) 
mesh_dir = pinocchio_model_dir 
urdf_model_path = model_path / "ur_description/urdf/ur5_robot.urdf" 
```

# Pinocchio Model Info

```python
robot = RobotWrapper.BuildFromURDF(urdf_model_path, mesh_dir)
model = robot.model
```

```python
print("=== Pinocchio Model Info ===")
print("Number of joints:", model.njoints)
print("Number of frames:", model.nframes)
print("nq:", model.nq)
print("nv:", model.nv)

print("\n=== Joint List ===")
for idx, joint in enumerate(model.joints):
    print(f"[{idx}] {joint}, nq={joint.nq}, nv={joint.nv}")

print("\n=== Frame List ===")
for idx, frame in enumerate(model.frames):
    print(f"[{idx}] {frame.name}, type={frame.type}, parent={frame.parent}")

print("\n=== Inertia List ===")
for idx, inertia in enumerate(model.inertias):
    print(f"[{idx}] inertia mass={inertia.mass}, com={inertia.lever}")
```

# Pinocchio Data Info

```python
data = robot.data
```

```python
print("=== Pinocchio Data Info ===")
print("\n=== Link Poses (oMi) ===")
for idx, Mi in enumerate(data.oMi):
    print(f"[{idx}] translation={Mi.translation.T}, rotation=\n{Mi.rotation}")

print("\n=== Mass Matrix (M) ===")
print(data.M)

print("\n=== Joint Velocities (v) ===")
for idx, vel in enumerate(data.v):
    print(f"Joint {idx}: {vel}")

print("\n=== Joint Accelerations (a) ===")
for idx, acc in enumerate(data.a):
    print(f"Joint {idx}: {acc}")

print("\n=== Center of Mass (CoM) ===")
print("CoM position:", data.com[0].T)
print("CoM velocity:", data.vcom[0].T)
```
